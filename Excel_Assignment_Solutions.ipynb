## 1. Data Cleaning

### a. What is data cleaning, and why is it important in data analysis?

Data cleaning is the process of finding and correcting inaccurate, incomplete, duplicate, or inconsistent data in a dataset.

It is important because clean data:
- improves the accuracy of analysis,
- reduces errors,
- makes reports more reliable,
- helps in better decision-making.

### b. Potential consequences of analyzing unclean or messy data

Unclean data may cause:
- incorrect results,
- misleading reports,
- duplicate records,
- wrong calculations,
- poor business decisions.

### c. Common steps involved in cleaning and organizing data

1. Remove duplicate records.
2. Handle missing or blank values.
3. Correct spelling and formatting errors.
4. Standardize dates, text, and number formats.
5. Remove unnecessary spaces.
6. Check for invalid or inconsistent values.
7. Arrange and organize the cleaned data properly.

## 3. Use of LEFT, RIGHT, MID, and CONCAT in Data Cleaning

### LEFT
Extracts characters from the beginning of a text.

```excel
=LEFT(A1,5)
```

Example: If A1 contains `ExcelFunctions`, the result will be `Excel`.

### RIGHT
Extracts characters from the end of a text.

```excel
=RIGHT(A1,4)
```

Example: If A1 contains `Report.xlsx`, the result will be `xlsx`.

### MID
Extracts characters from the middle of a text.

```excel
=MID(A1,6,4)
```

Example: If A1 contains `ExcelTipsAreGreat`, the result will be `Tips`.

### CONCAT
Combines text from two or more cells.

```excel
=CONCAT(A1," ",B1)
```

It is useful for joining first name and last name, combining codes, addresses, etc.

## 4. Role of Date Functions Like TODAY() in Managing Datasets

The `TODAY()` function returns the current date automatically.

```excel
=TODAY()
```

It is useful for:
- calculating age,
- finding the number of days remaining,
- tracking deadlines,
- checking overdue records,
- creating date-based reports.

The value updates automatically when the workbook is recalculated or reopened on another day.

## 5. Data Validation for the Quantity Column

The Quantity column should accept only whole numbers between **1 and 10**.

### a. Configure an input message

1. Select the cells in the **Quantity** column.
2. Go to **Data > Data Validation**.
3. Under **Settings**:
   - Allow: `Whole number`
   - Data: `between`
   - Minimum: `1`
   - Maximum: `10`
4. Open the **Input Message** tab.
5. Enter:

**Title:** Quantity Limit  
**Input Message:** `Please enter a whole number between 1 and 10.`

### b. Set an error alert

1. Open the **Error Alert** tab.
2. Select **Stop** as the style.
3. Enter:

**Title:** Invalid Input  
**Error Message:** `The quantity must be a whole number between 1 and 10.`

4. Click **OK**.

Now Excel will reject values below 1, above 10, or non-whole-number entries.

## 8. TODAY() and NOW() Functions

### a. Difference between TODAY() and NOW()

- `TODAY()` returns only the current **date**.
- `NOW()` returns the current **date and time**.

Examples:

```excel
=TODAY()
```

```excel
=NOW()
```

### b. Find how many days are left until 2025-06-10

If cell A1 contains `2025-06-10`:

```excel
=A1-TODAY()
```

This returns the number of days remaining until the date. If the date has already passed, the result will be negative.

### c. Display current date and time as MM/DD/YYYY HH:MM AM/PM

```excel
=TEXT(NOW(),"MM/DD/YYYY HH:MM AM/PM")
```

### d. What happens to =TODAY() when the worksheet is reopened the next day?

The value updates automatically and shows the new current date.

### e. Keyboard shortcut to enter today's date as a static value

```text
Ctrl + ;
```

This inserts today's date as a fixed value and it will not change automatically.
